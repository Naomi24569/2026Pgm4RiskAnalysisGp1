```python
import numpy as np

path = "2025_combined_mlar_header.txt"

selected_columns = [
    "state_code",
    "loan_type",
    "action_taken",
    "debt_to_income_ratio"
]

with open(path, "r", encoding="utf-8-sig") as f:
    header = f.readline().strip().split("|")

column_indices = [
    header.index(col) for col in selected_columns
]

extracted_data = np.genfromtxt(
    path,
    delimiter="|",
    skip_header=1,
    usecols=column_indices,
    max_rows=9998,
    dtype=str,
    encoding="utf-8"
)

print(extracted_data[:10])
print(extracted_data.shape)
print(type(extracted_data))
```

```python
import numpy as np

column_names = [
    "state",
    "loan_type",
    "action_taken",
    "debt_to_income"
]

missing_mask = np.isin(
    np.char.strip(extracted_data),
    ["", "NA", "nan", "NaN"]
)

for i, name in enumerate(column_names):
    print(name, "missing:", np.sum(missing_mask[:, i]))
```

```python
import numpy as np
import re

cleaned_data = extracted_data.copy()

def convert_dti(value):
    value = str(value).strip()

    if value in ("", "NA", "nan", "NaN"):
        return np.nan

    try:
        return float(value)
    except ValueError:
        pass

    match = re.fullmatch(
        r"(\d+(?:\.\d+)?)%-<(\d+(?:\.\d+)?)%",
        value
    )

    if match:
        lower = float(match.group(1))
        upper = float(match.group(2))
        return (lower + upper) / 2

    if value == ">60%":
        return 65.0

    return np.nan

dti_numeric = np.array([
    convert_dti(x) for x in cleaned_data[:, 3]
], dtype=float)

print("Missing before:",
      np.isnan(dti_numeric).sum())

indices = np.arange(len(dti_numeric))
valid = ~np.isnan(dti_numeric)
missing = np.isnan(dti_numeric)

if valid.sum() >= 2:
    dti_interpolated = dti_numeric.copy()

    inside = (
        missing &
        (indices > indices[valid][0]) &
        (indices < indices[valid][-1])
    )

    dti_interpolated[inside] = np.interp(
        indices[inside],
        indices[valid],
        dti_numeric[valid]
    )
else:
    dti_interpolated = dti_numeric.copy()

print("Missing after:",
      np.isnan(dti_interpolated).sum())

print(dti_interpolated[:10])
```

```python
other_columns = extracted_data[:, :3]

final_data = np.column_stack((
    other_columns,
    dti_interpolated
))

print(final_data[:10])
print(final_data.shape)
print(type(final_data))
```

```python
for i, name in enumerate([
    "state", "loan_type", "action_taken", "debt_to_income"
]):
    missing = np.isin(
        np.char.strip(final_data[:, i]),
        ["", "NA", "nan", "NaN"]
    )
    print(name, ":", np.sum(missing))
```

```python
import numpy as np

final_cleaned_data = final_data.copy()

state_missing = np.isin(
    np.char.strip(final_cleaned_data[:, 0]),
    ["", "NA", "nan", "NaN"]
)

final_cleaned_data[:, 0] = np.where(
    state_missing,
    "Unknown",
    final_cleaned_data[:, 0]
)

print("Unknown states:",
      np.sum(final_cleaned_data[:, 0] == "Unknown"))

print("Shape:", final_cleaned_data.shape)
print(type(final_cleaned_data))
```
