# C Module: DTI Grouping and Approval Rates
Development notebook using demo data.
Current task:
calculate the approval rate for one DTI group.
Definition:
Approved= action_taken 1 or 2.
Valid applications= action_taken 1,2 or 3.    

```python
actions=[1,2,3,1,4]
total=0
approved=0

for action in actions:
    if action in(1,2,3):
        total+=1
        if action in(1,2):
            approved+=1
print("Valid applications:",total)
print("Approved applications:",approved)

if total>0:
    approval_rate=approved/total
    print("Approved rate:",approval_rate*100,"%")
else:
    approved_rate=None
    print("No valid applications")
            
```

##Step2: Identify the DTI group
Classify a DTI value into one of six groups.

```python
# Demo DTI value storted as text
dti="40"
if dti in ["","NA","Exempt"]:
    group=None
elif dti=="<20%":
    group="<20%"
elif dti =="20%-30%":
    group="20%-30%"
elif dti=="30%-36%":
    group="30%-36%"
elif dti=="50%-60%":
    group="50%-60%"
elif dti==">60%":
    group=">60%"
else:
    dti_number=int(dti)
    if dti_number>=36 and dti_number<50:
        group="36%-50%"
    else:
        group=None
print("DTI:",dti)
print("Group:",group)
    
```
