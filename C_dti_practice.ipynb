# C Module: DTI Grouping and Approval Rates
Development notebook using demo data.
Current task:
calculate the approval rate for one DTI group.
Definition:
Approved= action_taken 1 or 2.
Valid applications= action_taken 1,2 or 3.    

```python
actions=[1,2,3,1,4]
total=0
approved=0

for action in actions:
    if action in(1,2,3):
        total+=1
        if action in(1,2):
            approved+=1
print("Valid applications:",total)
print("Approved applications:",approved)

if total>0:
    approval_rate=approved/total
    print("Approved rate:",approval_rate*100,"%")
else:
    approved_rate=None
    print("No valid applications")
            
```
