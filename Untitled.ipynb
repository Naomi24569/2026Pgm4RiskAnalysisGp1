```python
import numpy as np

saved_datasets = []
saved_datasets_v2 = {}
intermediate_results = []

state = "CA"
loan_type = "1"

dti_bins = np.array([
    "<20%", "20%-<30%", "30%-36%",
    "36%-<50%", "50%-60%", ">60%"
])

approval_rates = np.array([
    0.92, 0.85, 0.76, 0.61, 0.43, 0.25
])

size = np.array([
    120, 130, 160, 180, 220, 250
])

another_state = "TX"
another_loan_type = "1"

another_dti_bins = np.array([
    "<20%", "20%-<30%", "30%-36%",
    "36%-<50%", "50%-60%", ">60%"
])

another_approval_rates = np.array([
    0.81, 0.96, 0.55, 0.42, 0.68, 0.98
])

another_size = np.array([
    121, 132, 165, 185, 221,253
])
```

```python
class StateData:
    def __init__(self, state, loan_type, dti_bin, approval_rates, size):
        self.state = state  
        self.loan_type = loan_type 
        self.dti_bin = dti_bin 
        self.approval_rates = approval_rates
        self.size = size

    def print_data(self):
        print(f"state is {self.state}, Loan Type is {self.loan_type}, dti_bin is, {self.dti_bin}, Approval Rates is {self.approval_rates}, Size is {self.size}")
```

```python
def data_saver(state, loan_type, dti_bins, approval_rates, size):
    for i in range(len(dti_bins)):
        new_data = StateData(state, loan_type, dti_bins[i], approval_rates[i], size[i])
        new_data_hash = hash((state, loan_type, dti_bins[i]))
        
        if (new_data_hash in saved_datasets_v2):
            overideYN = input("You are trying to save an existing record, you sure want to overide? Y/N")
            if(overideYN == "Y" or overideYN == "y"):
                saved_datasets_v2[new_data_hash] = new_data
                new_data.print_data()
        else:
            saved_datasets_v2[new_data_hash] = new_data
            new_data.print_data()
```

```python
def comparison_analysis(state_1, state_2, loan_type):
    state_1_datasets = []
    state_2_datasets = []
    for state_data in saved_datasets:
        if state_data.state == state_1 and state_data.loan_type == loan_type:
            state_1_datasets.append(state_data)
        elif state_data.state == state_2 and state_data.loan_type == loan_type:
            state_2_datasets.append(state_data)
```

```python
data_saver(another_state, another_loan_type, another_dti_bins, another_approval_rates, size)
data_saver(state, loan_type, dti_bins, approval_rates, another_size)
```
